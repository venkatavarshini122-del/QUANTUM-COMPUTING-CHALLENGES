```python
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Initialize simulator
simulator = AerSimulator()

# ==============================================================================
# 1. CIRCUIT 1: Biased Towards |0⟩ (Bonus Challenge)
# ==============================================================================
# Single-qubit circuit with measurement
qc_bias_0 = QuantumCircuit(1, 1)

# Apply Ry rotation with theta = pi/3 (60 degrees)
# State vector: cos(pi/6)|0⟩ + sin(pi/6)|1⟩
qc_bias_0.ry(np.pi / 3, 0)
qc_bias_0.measure(0, 0)

# Predictions for Circuit 1
# P(0) = cos^2(pi/6) = 0.75 (75%)
# P(1) = sin^2(pi/6) = 0.25 (25%)
pred_0 = {"P(0)": "75%", "P(1)": "25%"}

# Execute Circuit 1 for 100 shots
compiled_qc0 = transpile(qc_bias_0, simulator)
job_0 = simulator.run(compiled_qc0, shots=100)
counts_0 = job_0.result().get_counts()


# ==============================================================================
# 2. CIRCUIT 2: Biased Towards |1⟩ (Bonus Challenge)
# ==============================================================================
# Single-qubit circuit with measurement
qc_bias_1 = QuantumCircuit(1, 1)

# Apply Ry rotation with theta = 2*pi/3 (120 degrees)
# State vector: cos(pi/3)|0⟩ + sin(pi/3)|1⟩
qc_bias_1.ry(2 * np.pi / 3, 0)
qc_bias_1.measure(0, 0)

# Predictions for Circuit 2
# P(0) = cos^2(pi/3) = 0.25 (25%)
# P(1) = sin^2(pi/3) = 0.75 (75%)
pred_1 = {"P(0)": "25%", "P(1)": "75%"}

# Execute Circuit 2 for 100 shots
compiled_qc1 = transpile(qc_bias_1, simulator)
job_1 = simulator.run(compiled_qc1, shots=100)
counts_1 = job_1.result().get_counts()


# ==============================================================================
# 3. PRINT CIRCUIT DIAGRAMS, PREDICTIONS, AND MEASUREMENT RESULTS
# ==============================================================================
print("=" * 70)
print("CIRCUIT 1: BIASED TOWARDS |0⟩")
print("=" * 70)
print("Quantum Circuit Diagram:")
print(qc_bias_0.draw('text'))
print(f"\nPredicted Probabilities: {pred_0}")
print(f"Measurement Results (100 shots): {counts_0}\n")

# Text-based Histogram for Circuit 1
c0_0 = counts_0.get('0', 0)
c0_1 = counts_0.get('1', 0)
print("Histogram (Text Representation):")
print(f"  |0⟩ : {'█' * (c0_0 // 2)} ({c0_0} shots)")
print(f"  |1⟩ : {'█' * (c0_1 // 2)} ({c0_1} shots)\n")

print("=" * 70)
print("CIRCUIT 2: BIASED TOWARDS |1⟩")
print("=" * 70)
print("Quantum Circuit Diagram:")
print(qc_bias_1.draw('text'))
print(f"\nPredicted Probabilities: {pred_1}")
print(f"Measurement Results (100 shots): {counts_1}\n")

# Text-based Histogram for Circuit 2
c1_0 = counts_1.get('0', 0)
c1_1 = counts_1.get('1', 0)
print("Histogram (Text Representation):")
print(f"  |0⟩ : {'█' * (c1_0 // 2)} ({c1_0} shots)")
print(f"  |1⟩ : {'█' * (c1_1 // 2)} ({c1_1} shots)\n")


# ==============================================================================
# 4. DISPLAY GRAPHICAL HISTOGRAMS OF RESULTS
# ==============================================================================
# Individual Histogram for Circuit 1
plot_histogram(counts_0, title="Circuit 1 Measurement Results (Biased towards |0⟩)")
plt.show()

# Individual Histogram for Circuit 2
plot_histogram(counts_1, title="Circuit 2 Measurement Results (Biased towards |1⟩)")
plt.show()

# Comparative Histogram for Both Circuits
plot_histogram(
    [counts_0, counts_1], 
    legend=['Biased towards |0⟩', 'Biased towards |1⟩'], 
    title="Comparison of Biased Quantum Coin Measurement Results (100 Shots)"
)
plt.show()


# ==============================================================================
# 5. EXPLANATION OF GATE IMPACT
# ==============================================================================
explanation = """
================================================================================
EXPLANATION OF GATE IMPACT:
================================================================================
By using parameterised single-qubit rotation gates such as Ry(θ), we rotate the 
qubit's state vector along the Y-axis of the Bloch sphere. 

The resulting state is |ψ⟩ = cos(θ/2)|0⟩ + sin(θ/2)|1⟩.

- For θ = π/3 (Circuit 1): 
  The angle θ/2 = π/6, yielding probability amplitudes cos(π/6) ≈ 0.866 for |0⟩ 
  and sin(π/6) = 0.5 for |1⟩. This translates to P(0) = 75% and P(1) = 25%.

- For θ = 2π/3 (Circuit 2): 
  The angle θ/2 = π/3, yielding probability amplitudes cos(π/3) = 0.5 for |0⟩ 
  and sin(π/3) ≈ 0.866 for |1⟩. This translates to P(0) = 25% and P(1) = 75%.
================================================================================
"""

print(explanation)
```

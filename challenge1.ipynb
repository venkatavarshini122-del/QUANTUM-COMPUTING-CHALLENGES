```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# Setup backend and constraints (100 shots)
backend = AerSimulator()
shots = 100

# ==============================================================================
# CIRCUIT A: Apply 1 X Gate
# ==============================================================================
qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)

counts_a = backend.run(qc_a, shots=shots).result().get_counts()

print("--- CIRCUIT A ---")
print("Prediction: {'1': 100}")
print("Actual Results:", counts_a)
display(qc_a.draw('mpl'))
display(plot_histogram(counts_a, title="Circuit A Results"))


# ==============================================================================
# CIRCUIT B: Apply 2 X Gates
# ==============================================================================
qc_b = QuantumCircuit(1, 1)
qc_b.x(0)
qc_b.x(0)
qc_b.measure(0, 0)

counts_b = backend.run(qc_b, shots=shots).result().get_counts()

print("\n--- CIRCUIT B ---")
print("Prediction: {'0': 100}")
print("Actual Results:", counts_b)
display(qc_b.draw('mpl'))
display(plot_histogram(counts_b, title="Circuit B Results"))


# ==============================================================================
# CIRCUIT C: Apply 1 H Gate
# ==============================================================================
qc_c = QuantumCircuit(1, 1)
qc_c.h(0)
qc_c.measure(0, 0)

counts_c = backend.run(qc_c, shots=shots).result().get_counts()

print("\n--- CIRCUIT C ---")
print("Prediction: ~50% '0' and ~50% '1' (~50 counts each)")
print("Actual Results:", counts_c)
display(qc_c.draw('mpl'))
display(plot_histogram(counts_c, title="Circuit C Results"))


# ==============================================================================
# SUBMISSION: SHORT COMPARISON & ANSWERS
# ==============================================================================
print("\n" + "="*70)
print("SUBMISSION EXPLANATION & COMPARISON")
print("="*70)
print("""
1. PREDICTIONS VS ACTUAL RESULTS COMPARISON:
   - Circuit A (1 X Gate): Predicted 100% '1'. Actual matches prediction 
     completely because the X gate deterministically flips state |0> to |1>.
   - Circuit B (2 X Gates): Predicted 100% '0'. Actual matches prediction 
     completely because two X gates cancel out (|0> -> |1> -> |0>).
   - Circuit C (1 H Gate): Predicted ~50 counts of '0' and ~50 counts of '1'. 
     Actual results closely follow this split with slight variations due to 
     quantum randomness and measurement sampling noise over 100 shots.


""")
```
